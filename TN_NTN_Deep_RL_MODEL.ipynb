```python
Hybrid_Network = pd.read_csv('/content/Hybrid_Network_TN_NTN.csv')
Hybrid_Network
```

```python
# توحيد أسماء الشبكات
Hybrid_Network["network_type"] = Hybrid_Network["network_type"].replace({
    "5G_NR": "NR_5G",
    "WiFi6_5GHz": "WiFi",
    "WiFi6_2.4GHz": "WiFi",
    "WiFi6E_6GHz": "WiFi",
    "LEO": "SAT (LEO)"
})
```

```python
Hybrid_Network
```

```python
Hybrid_Network['network_type'].value_counts()
```

```python
import numpy as np

# =====================================================
# Allowed Areas for each Network
# =====================================================

network_area = {

    "NR_5G": {
        "areas": ["Indoor", "Urban", "Rural", "Highway"],
        "p":     [0.41,      0.30,     0.18,    0.11]
    },

    "WiFi": {
        "areas": ["Indoor", "Urban", "Rural"],
        "p":     [0.47,      0.33,     0.20]
    },

    "SAT (LEO)": {
        "areas": ["Indoor", "Urban", "Rural", "Highway", "Maritime", "Desert"],
        "p":     [0.35,      0.25,     0.15,     0.10,      0.08,      0.07]
    },

    "HAPS": {
        "areas": ["Urban", "Rural", "Highway", "Maritime", "Desert"],
        "p":     [0.39,      0.23,      0.15,      0.12,      0.11]
    },

    "UAV": {
        "areas": ["Urban", "Rural", "Highway", "Maritime", "Desert"],
        "p":     [0.38,      0.22,      0.16,      0.13,      0.11]
    }

}

# =====================================================
# Assign Area according to Network Type
# =====================================================

def assign_area(network):

    info = network_area[network]

    return np.random.choice(
        info["areas"],
        p=info["p"]
    )

Hybrid_Network["Area"] = Hybrid_Network["network_type"].apply(assign_area)

# =====================================================
# Available Networks in each Area
# =====================================================

area_networks = {

    "Indoor": "NR_5G,WiFi,SAT (LEO)",

    "Urban": "NR_5G,WiFi,UAV,HAPS,SAT (LEO)",

    "Rural": "NR_5G,WiFi,HAPS,SAT (LEO),UAV",

    "Highway": "NR_5G,UAV,HAPS,SAT (LEO)",

    "Maritime": "UAV,HAPS,SAT (LEO)",

    "Desert": "UAV,HAPS,SAT (LEO)"

}

Hybrid_Network["Available_Networks"] = Hybrid_Network["Area"].map(area_networks)

# =====================================================
# Check Results
# =====================================================

print(Hybrid_Network["Area"].value_counts())

print("\n----------------------------\n")

print(
    Hybrid_Network.groupby(
        ["Area", "network_type"]
    ).size()
)
```

```python
import numpy as np

# إنشاء 20 User IDs
user_ids = [f"UE_{i:03d}" for i in range(20)]

# اختيار صفوف WiFi
wifi_mask = Hybrid_Network["network_type"] == "WiFi"

# إعادة توزيع الـ User IDs على بيانات WiFi
Hybrid_Network.loc[wifi_mask, "UE_ID"] = np.resize(
    user_ids,
    wifi_mask.sum()
)

# خلط الـ IDs حتى لا تكون مرتبة
Hybrid_Network.loc[wifi_mask, "UE_ID"] = np.random.permutation(
    Hybrid_Network.loc[wifi_mask, "UE_ID"].values
)

# التحقق
print(
    Hybrid_Network.groupby("network_type")["UE_ID"].nunique()
)
```

```python
Hybrid_Network["UE_ID"].value_counts()
```

```python
import re

def normalize_user_id(uid):
    # استخراج الرقم من أي ID
    number = re.findall(r'\d+', str(uid))[0]
    return f"UE_{int(number):03d}"

Hybrid_Network["UE_ID"] = Hybrid_Network["UE_ID"].apply(normalize_user_id)
```

```python
print(Hybrid_Network["UE_ID"].nunique())
```

```python
print(
    Hybrid_Network.groupby("network_type")["UE_ID"].nunique()
)
```

```python
# تحويل UE_000 -> User_01

Hybrid_Network["UE_ID"] = (
    Hybrid_Network["UE_ID"]
    .str.extract(r'(\d+)')[0]
    .astype(int)
    .apply(lambda x: f"User_{x+1:02d}")
)

print(Hybrid_Network["UE_ID"].unique())
```
